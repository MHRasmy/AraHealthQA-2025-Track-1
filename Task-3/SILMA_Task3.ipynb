````python
# pip install -U huggingface_hub transformers
import os
from types import SimpleNamespace
from huggingface_hub import InferenceClient
from huggingface_hub.utils import HfHubHTTPError
from transformers import AutoTokenizer

HF_TOKEN = os.getenv("HF_TOKEN")
MODEL_ID = "silma-ai/SILMA-Kashif-2B-Instruct-v1.0"   # or: "silma-ai/SILMA-Kashif-2B-Instruct-v1.0"

def _to_chat_list(messages):
    if isinstance(messages, str):
        return [{"role": "user", "content": messages}]
    chat = []
    for role, content in messages:
        role = {"human": "user", "ai": "assistant"}.get(role, role)
        chat.append({"role": role, "content": content})
    return chat

class SilmaHFEndpoint:
    def __init__(self, model_id=MODEL_ID, temperature=0.0, max_new_tokens=500, timeout=120, provider=None):
        # provider can be None (auto), "featherless-ai", or "hf-inference"
        self.client = InferenceClient(model=model_id, token=HF_TOKEN, timeout=timeout, provider=provider)
        self.tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=True)
        self.temperature = temperature
        self.max_new_tokens = max_new_tokens

    def invoke(self, messages):
        chat = _to_chat_list(messages)

        # 1) Preferred path: conversational/chat API
        try:
            out = self.client.chat_completion(
                messages=[{"role": m["role"], "content": m["content"]} for m in chat],
                max_tokens=self.max_new_tokens,
                temperature=self.temperature,
            )
            # HF returns choices; message can be dict or object depending on version
            msg = out.choices[0].message
            text = (msg["content"] if isinstance(msg, dict) else msg.content) or ""
            return SimpleNamespace(content=text.strip())
        except Exception as e_chat:
            # 2) Fallback: text_generation with chat template (if provider supports it)
            prompt = self.tokenizer.apply_chat_template(chat, tokenize=False, add_generation_prompt=True)
            try:
                text = self.client.text_generation(
                    prompt,
                    max_new_tokens=self.max_new_tokens,
                    temperature=self.temperature,
                    do_sample=self.temperature > 0,
                    return_full_text=False,
                )
                return SimpleNamespace(content=text.strip())
            except Exception as e_tg:
                raise RuntimeError(f"Both chat and text-generation calls failed: chat={e_chat} | text-gen={e_tg}")

# swap into your pipeline:
# llm = ChatGroq(...)
llm = SilmaHFEndpoint(MODEL_ID, temperature=0.0, max_new_tokens=500)
````

````python
def check_local_knowledge(query, context):
    """Router function to determine if we can answer from local knowledge"""
    prompt = '''Role: Question-Answering Assistant
Task: Determine whether the system can answer the user's question based on the provided text.
Instructions:
    - Analyze the text and identify if it contains the necessary information to answer the user's question.
    - Provide a clear and concise response indicating whether the system can answer the question or not.
    - Your response should include only a single word. Nothing else, no other text, information, header/footer.
Output Format:
    - Answer: Yes/No
Study the below examples and based on that, respond to the last question.
Examples:
    Input:
        Text: The dose of this medicine will be different for different patients. Follow your doctor's orders or the directions on the label. The following information includes only the average doses of this medicine. If your dose is different, do not change it unless your doctor tells you to do so. Children 10 to 17 years of age—At first, 0.5 mg once a day, in the morning or evening. Your doctor may adjust your dose as needed. However, the dose is usually not more than 6 mg per day.
        User Question: لدي بنت عندها تاخر في النمو العقلي وتاخد دوا واحد بس اسمو رسبيدرال ما بعرف قديش اعطيها حبه بليوم لاني بطلت اتابع فيها عند الدكتور من زمان وهسه عمرها ١٣ سنه وضعيفه
    Expected Output:
        Answer: Yes
    Input:
        Text: The episode is not attributable to the physiological effects of a substance or another medical condition. Note: Criteria A–C represent a major depressive episode. Note: Responses to a significant loss (e.g., bereavement, financial ruin, losses from a natural disaster, a serious medical illness or disability) may include the feelings of intense sadness, rumination about the loss, insomnia, poor appetite, and weight loss noted in Criterion A, which may resemble a depressive episode.
        User Question: لدي اكتئاب مفاجئ بعد أن تعرضت لصوت رعد مفاجئ وأمطار وأصبحت كئيية لمدة ثلاث ايام وأبكي بدون سبب هل هذه صدمة ام فقط فترة ثم قلت انه فقط اكتئاب المطر حتى أن اشرقت الشمس قد لازالت هذه الحالة معي لم اعرف السبب واحس كأنه هناك ضغط في صدري
    Expected Output:
        Answer: No
    Input:
        User Question: {query}
        Text: {text}
'''
    formatted_prompt = prompt.format(text=context, query=query)
    response = llm.invoke(formatted_prompt)
    return response.content.strip().lower() == "yes"
````

````python
from crewai_tools import SerperDevTool, ScrapeWebsiteTool

def setup_web_scraping_agent():
    """Setup the web scraping agent and related components"""
    search_tool = SerperDevTool()  # Tool for performing web searches
    scrape_website = ScrapeWebsiteTool()  # Tool for extracting data from websites

    # Define the web search agent
    web_search_agent = Agent(
        role="Expert Trusted‑Source Hunter",
        goal=(
          "For a given mental‑health question, locate ONE high‑quality, freely "
          "accessible webpage (Mayo Clinic, NIH, WHO, PubMed, etc.) that contains "
          "the *core context* needed to answer in the requested style "
          "(information / direct guidance / emotional support)."
        ),
        backstory=(
        "You are a specialised researcher who only relies on reputable health "
        "sites. You never choose forums, commercial blogs, or low‑authority pages."
        ),
        allow_delegation=False,
        verbose=True,
        llm=crew_llm
    )

    # Define the web scraping agent
    web_scraper_agent = Agent(
        role="Medical‑Context Extractor",
        goal=(
            "Download the page identified by the search agent and pull out the "
            "key paragraphs that supply the missing medical context. "
            "Return a clean JSON with title, url, and a concise Arabic summary."
        ),
        backstory=(
            "You have deep experience stripping boilerplate and focusing on medical "
            "facts, guidelines, or supportive language that matches the needed style."
        ),
        allow_delegation=False,
        verbose=True,
        llm=crew_llm,
    )

    # Define the web search task
    search_task = Task(
        description=(
            "Identify the most relevant web page or article for the topic: '{topic}'. "
            "Use all available tools to search for and provide a link to a web page "
            "that contains valuable information about the topic. Keep your response concise."
        ),
        expected_output=(
            "A concise summary of the most relevant web page or article for '{topic}', "
            "including the link to the source and key points from the content."
        ),
        tools=[search_tool],
        agent=web_search_agent,
    )

    # Define the web scraping task
    scraping_task = Task(
        description=(
            "Extract and analyze data from the given web page or website. Focus on the key sections "
            "that provide insights into the topic: '{topic}'. Use all available tools to retrieve the content, "
            "and summarize the key findings in a concise manner."
        ),
        expected_output=(
            "A detailed summary of the content from the given web page or website, highlighting the key insights "
            "and explaining their relevance to the topic: '{topic}'. Ensure clarity and conciseness."
        ),
        tools=[scrape_website],
        agent=web_scraper_agent,
    )

    # Define the crew to manage agents and tasks
    crew = Crew(
        agents=[web_search_agent, web_scraper_agent],
        tasks=[search_task, scraping_task],
        verbose=1,
        memory=False,
    )
    return crew

def get_web_content(query):
    """Get content from web scraping"""
    crew = setup_web_scraping_agent()
    result = crew.kickoff(inputs={"topic": query})
    return result.raw
````

````python
def setup_vector_db(pdf_path):
    """Setup vector database from PDF"""
    # Load and chunk PDF
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=50
    )
    chunks = text_splitter.split_documents(documents)

    # Create vector database
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-mpnet-base-v2"
    )
    vector_db = FAISS.from_documents(chunks, embeddings)

    return vector_db

def get_local_content(vector_db, query):
    """Get content from vector database"""
    docs = vector_db.similarity_search(query, k=5)
    return " ".join([doc.page_content for doc in docs])
````

````python
def generate_final_answer(context, query):
    """Generate final answer using LLM"""
    # messages = [
    #     (
    #         "user",
    #         f"You are **AraHealthQA Assistant**, a large‑language model helping Arabic speakers with mental‑health‑related questions. Use the provided context to answer the query accurately. Keep your answer ≤ 200 Arabic words unless user explicitly asked for detail.\n\nContext:\n {context}\n\nQuery:\n{query}"
    #     ),
    #     ("system", f"Context: {context}"),
    #     ("human", query),
    # ]
    messages = f"You are **AraHealthQA Assistant**, a large‑language model helping Arabic speakers with mental‑health‑related questions. Use the provided context to answer the query accurately. Keep your answer ≤ 200 Arabic words unless user explicitly asked for detail.\n\nContext:\n {context}\n\nQuery:\n{query}"
    response = llm.invoke(messages)
    return response.content

def process_query(query, vector_db, local_context):
    """Main function to process user query"""
    print(f"Processing query: {query}")

    # Step 1: Check if we can answer from local knowledge
    can_answer_locally = check_local_knowledge(query, local_context)
    print(f"Can answer locally: {can_answer_locally}")

    # Step 2: Get context either from local DB or web
    if can_answer_locally:
        context = get_local_content(vector_db, query)
        print("Retrieved context from local documents")
    else:
        context = get_web_content(query)
        print("Retrieved context from web scraping")

    # Step 3: Generate final answer
    answer = generate_final_answer(context, query)
    return answer
````

````python
# -*- coding: utf-8 -*-
"""Agentic RAG pipeline for AraHealthQA Track‑1 (Sub‑task 3)
=============================================================

This script wires together:
  • A Chroma vector DB already populated with English chunks (see `CHROMA_DIR`).
  • Qwen‑3‑8B (4‑bit NF4) for generation.
  • A simple router that inspects the classification labels → decides retrieval mode & prompt style.

It exposes one public call:
    answer_question(question: str, predicted_labels: list[str]) -> str

Author: ChatGPT 2025‑07‑27
"""

from __future__ import annotations

import os
from pathlib import Path
from typing import List, Tuple, Dict, Any

from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    pipeline,
)
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.docstore.document import Document
from langchain.text_splitter import RecursiveCharacterTextSplitter

# ────────────────────────────────────────────────────────────
# 1.  CONFIG ─────────────────────────────────────────────────
# ────────────────────────────────────────────────────────────

MODEL_ID = "Qwen/Qwen3-8B"
# CHROMA_DIR = "/content/chroma_index"  # change to your path
DEVICE = 0  # set to -1 for CPU

# Category → behaviour policy
a_CategoryPolicy = Dict[str, Any]
CATEGORY_POLICIES: Dict[Any, a_CategoryPolicy] = {
    # Single‑label policies
    "information": {
        "style_key": "expository",
        "retrieval": "always",
        "k": 6,
    },
    "direct_guidance": {
        "style_key": "action_oriented",
        "retrieval": "hybrid",
        "k": 4,
    },
    "emotional_support": {
        "style_key": "empathetic",
        "retrieval": "avoid",
        "k": 0,
    },
    # Common multi‑label combinations (tuple keys)
    ("information", "direct_guidance"): {
        "style_key": "mix_expository_guidance",
        "retrieval": "always",
        "k": 6,
    },
    ("information", "emotional_support"): {
        "style_key": "expository_then_empathy",
        "retrieval": "always",
        "k": 6,
    },
    ("direct_guidance", "emotional_support"): {
        "style_key": "supportive_guidance",
        "retrieval": "hybrid",
        "k": 4,
    },
    ("information", "direct_guidance", "emotional_support"): {
        "style_key": "info_guidance_empathy",
        "retrieval": "always",
        "k": 6,
    },
}

# Style‑tag snippets inserted into the prompt for the LLM
STYLE_TAGS: Dict[str, str] = {
    "expository": "🔹FORMAT: factual answer · cite sources (1,2,3)…",
    "action_oriented": "🔹FORMAT: numbered steps · imperative verbs",
    "empathetic": "🔹FORMAT: warm tone · first acknowledge feelings",
    "mix_expository_guidance": "🔹FORMAT: facts first ↔ actionable steps second",
    "expository_then_empathy": "🔹FORMAT: factual answer then reassuring note",
    "supportive_guidance": "🔹FORMAT: gentle encouragement while giving steps",
    "info_guidance_empathy": "🔹FORMAT: facts → steps → empathy close",
}

# Prompt skeleton shown to the LLM
PROMPT_TEMPLATE = """
You are **AraHealthQA Assistant**, a large‑language model helping Arabic speakers
with mental‑health‑related questions.

**User Question**
{{question}}

**Predicted Answer Category (multi‑label)**
{{labels}}

**Retrieval Mode**
{{retrieval_instructions}}

{% if context %}
**CONTEXT (retrieved)**
{{context}}
{% endif %}

**Write the best possible answer following ALL rules below**
1. Obey Arabic medical‑ethics: no diagnoses or prescriptions → refer to professionals.
2. Follow style guide: {{style_tag}}
3. If you cite, enumerate sources as (1), (2)… matching order in CONTEXT.
4. If you lack enough info, say so briefly and suggest next steps.
5. Keep answer ≤ 200 Arabic words unless user explicitly asked for detail.
"""

# ────────────────────────────────────────────────────────────
# 2.  LIGHTWEIGHT UTILS ─────────────────────────────────────
# ────────────────────────────────────────────────────────────


def needs_evidence(question: str) -> bool:
    """Heuristic: check if guidance likely needs external facts."""
    guidance_kw = [
        "how",
        "طريقة",
        "خطوات",
        "step",
        "method",
        "best way",
    ]
    q_lower = question.lower()
    return any(k.lower() in q_lower for k in guidance_kw)


def get_retrieval_instructions(mode: str) -> str:
    if mode == "always":
        return (
            "Use the CONTEXT section. If insufficient, say “I need more info.”"
        )
    if mode == "hybrid":
        return (
            "Use CONTEXT **if present**; otherwise rely on your own knowledge, but prefer "
            "evidence from CONTEXT when available."
        )
    return "Do NOT quote or rely on external sources; answer from your own knowledge."


# ────────────────────────────────────────────────────────────
# 3.  INITIALISE COMPONENTS ─────────────────────────────────
# ────────────────────────────────────────────────────────────

print("[INIT] Loading Qwen‑3‑8B (4‑bit)…")

# 3.1  LLM
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="float16",
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    device_map="auto" if DEVICE >= 0 else None,
    trust_remote_code=True,
    quantization_config=bnb_config,
)
````

````python
# -*- coding: utf-8 -*-
"""Agentic RAG pipeline for AraHealthQA Track‑1 (Sub‑task 3) with batch classification
==============================================================================

This script:
  • Loads a fine‑tuned multi‑label classifier to predict answer categories.
  • Reads the entire Subtask3_input_test.tsv, predicts categories for each question.
  • Retrieves context via Chroma + HuggingFace embeddings according to policy, filtering by similarity ≥ 0.6.
  • Includes few‑shot examples for emotional_support (no‑retrieval) from Train_Dev.tsv.
  • Generates Arabic answers with Qwen-3-8B (4‑bit NF4), tailored per category.
  • Optionally, precomputes prompts only (for retrieval-heavy scenarios) or runs full retrieval+generation batched.
  • Saves predictions & answers/prompts to CSV.

Usage:
  python rag_router.py --test-file /mnt/data/Subtask3_input_test.tsv \
                       [--train-file /mnt/data/Train_Dev.tsv] \
                       [--output-file output.csv] \
                       [--prompts-only]  # to generate prompts and skip generation
"""

from __future__ import annotations
import os
import argparse
from typing import List, Dict, Any

import torch
import pandas as pd
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    pipeline,
    AutoModelForSequenceClassification,
)
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.docstore.document import Document

# ──────────────────────────────────────────────────────
# 1. CONFIG
# ──────────────────────────────────────────────────────
MODEL_ID = "Qwen/Qwen3-8B"
EMBED_MODEL_ID = "Qwen/Qwen3-Embedding-4B"
DEVICE = 0  # GPU id; -1 for CPU

CLASSIFIER_MODEL_ID = "path/to/your/classifier"
CLASS_LABELS = ["information", "direct_guidance", "emotional_support"]
CLASS_THRESHOLD = 0.5

CHROMA_DIR = "/content/chroma_index"
SIMILARITY_THRESHOLD = 0.6

CATEGORY_POLICIES: Dict[Any, Dict[str, Any]] = {
    "information":       {"style_key": "expository",               "retrieval": "always", "k": 6},
    "direct_guidance":   {"style_key": "action_oriented",         "retrieval": "hybrid", "k": 4},
    "emotional_support": {"style_key": "empathetic",              "retrieval": "avoid",  "k": 0},
    ("information","direct_guidance"):            {"style_key": "mix_expository_guidance", "retrieval": "always", "k": 6},
    ("information","emotional_support"):           {"style_key": "expository_then_empathy","retrieval": "always", "k": 6},
    ("direct_guidance","emotional_support"):       {"style_key": "supportive_guidance",     "retrieval": "hybrid", "k": 4},
    ("information","direct_guidance","emotional_support"): {"style_key": "info_guidance_empathy", "retrieval": "always", "k": 6},
}
STYLE_TAGS = {
    "expository":            "🔹FORMAT: factual answer · cite sources (1,2,3)…",
    "action_oriented":       "🔹FORMAT: numbered steps · imperative verbs",
    "empathetic":            "🔹FORMAT: warm tone · first acknowledge feelings",
    "mix_expository_guidance":"🔹FORMAT: facts first ↔ actionable steps second",
    "expository_then_empathy":"🔹FORMAT: factual answer then reassuring note",
    "supportive_guidance":   "🔹FORMAT: gentle encouragement while giving steps",
    "info_guidance_empathy": "🔹FORMAT: facts → steps → empathy close",
}

PROMPT_TEMPLATE = """
{% if few_shot_block %}# Few-Shot Emotional Support Examples
${few_shot_block}

{% endif %}You are **AraHealthQA Assistant**, a large-language model helping Arabic speakers with mental-health-related questions.

**User Question**
${question}

**Predicted Answer Category (multi-label)**
${labels}

**Retrieval Mode**
${retrieval_instructions}

${context_block}
**Write the best possible answer following ALL rules below**
1. Obey Arabic medical-ethics: no diagnoses or prescriptions → refer to professionals.
2. Follow style guide: ${style_tag}
3. If you cite, enumerate sources as (1), (2)… matching order in CONTEXT.
4. If you lack enough info, say so briefly and suggest next steps.
5. Keep answer ≤ 200 Arabic words unless user explicitly asked for detail.
"""

# ──────────────────────────────────────────────────────
# 2. UTILS
# ──────────────────────────────────────────────────────

def needs_evidence(question: str) -> bool:
    kws = ["how","طريقة","خطوات","step","method","best way"]
    q = question.lower()
    return any(k in q for k in kws)


def get_retrieval_instructions(mode: str) -> str:
    return {
        "always": "Use CONTEXT (retrieved) & cite it; if insufficient say ‘I need more info.’",
        "hybrid": "Use CONTEXT if above threshold; otherwise rely on own knowledge, but prefer evidence.",
        "avoid":  "Do NOT quote or rely on external sources; answer from your own knowledge.",
    }[mode]

# ──────────────────────────────────────────────────────
# 3. LOAD MODELS & FEW-SHOT
# ──────────────────────────────────────────────────────
print("[INIT] Loading classification model…")
cls_tok = AutoTokenizer.from_pretrained(CLASSIFIER_MODEL_ID, trust_remote_code=True)
cls_mod = AutoModelForSequenceClassification.from_pretrained(
    CLASSIFIER_MODEL_ID, trust_remote_code=True
).to(DEVICE if DEVICE>=0 else "cpu")
cls_mod.eval()

# Load Train_Dev for emotional support examples
few_shot_block = ""
if os.path.exists(args.train_file if 'args' in globals() else 'Train_Dev.tsv'):
    df_dev = pd.read_csv(args.train_file if 'args' in globals() else 'Train_Dev.tsv', sep='\t', names=["question","answer","final_AS"])
    es = df_dev[df_dev['final_AS'].apply(lambda x: eval(x)==['3'])].head(2)
    pairs = es[['question','answer']].to_dict('records')
    few_shot_block = "\n\n".join([f"Q: {p['question']}\nA: {p['answer']}" for p in pairs])

print("[INIT] Loading Qwen-3-8B generation model…")
bnb = BitsAndBytesConfig(
    load_in_4bit=True, bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4", bnb_4bit_compute_dtype="float16",
)
gen_tok = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
gen_mod = AutoModelForCausalLM.from_pretrained(
    MODEL_ID, device_map="auto" if DEVICE>=0 else None,
    trust_remote_code=True, quantization_config=bnb,
)
gen_pipe = pipeline("text-generation", model=gen_mod, tokenizer=gen_tok,
                    device=DEVICE, max_new_tokens=512)

# ──────────────────────────────────────────────────────
# 4. SETUP RETRIEVER (SIMILARITY FILTER)
# ──────────────────────────────────────────────────────
print("[INIT] Connecting to Chroma…")
embed_fn = HuggingFaceEmbeddings(model_name=EMBED_MODEL_ID)
vdb = Chroma(persist_directory=CHROMA_DIR, embedding_function=embed_fn)

class Retriever:
    def __init__(self, db): self.db = db
    def search(self, q: str, k: int) -> List[str]:
        docs_scores = self.db.similarity_search_with_score(q, k=k)
        filtered = [d for d,s in docs_scores if s>=SIMILARITY_THRESHOLD]
        return [f"({i+1}) {d.page_content.replace(chr(10),' ').strip()}"
                for i,d in enumerate(filtered)]

retriever = Retriever(vdb)

# ──────────────────────────────────────────────────────
# 5. CORE FUNCTIONS
# ──────────────────────────────────────────────────────

def predict_labels(question: str) -> List[str]:
    inputs = cls_tok(question, return_tensors="pt", truncation=True).to(DEVICE if DEVICE>=0 else "cpu")
    with torch.no_grad(): logits = cls_mod(**inputs).logits[0]
    probs = torch.sigmoid(logits).cpu().tolist()
    sel = [lab for lab,p in zip(CLASS_LABELS, probs) if p>=CLASS_THRESHOLD]
    if not sel:
        sel = [CLASS_LABELS[int(torch.argmax(logits))]]
    return sel

from string import Template

def build_prompt(question: str, labels: List[str]) -> str:
    key = tuple(sorted(labels))
    policy = CATEGORY_POLICIES.get(key) or CATEGORY_POLICIES[labels[0]]
    # retrieval
    context = []
    if policy['retrieval']=="always" or (policy['retrieval']=="hybrid" and needs_evidence(question)):
        context = retriever.search(question, k=policy['k'])
    context_block = f"**CONTEXT (retrieved)**\n{'\n\n'.join(context)}" if context else ""
    tpl = Template(PROMPT_TEMPLATE)
    return tpl.substitute(
        few_shot_block=few_shot_block if policy['retrieval']=='avoid' else "",
        question=question,
        labels=", ".join(labels),
        retrieval_instructions=get_retrieval_instructions(policy['retrieval']),
        context_block=context_block,
        style_tag=STYLE_TAGS[policy['style_key']],
    )

def answer_question(question: str, labels: List[str]) -> str:
    prompt = build_prompt(question, labels)
    out = gen_pipe(prompt)[0]['generated_text']
    return out[len(prompt):].strip()

# ──────────────────────────────────────────────────────
# 6. BATCH PROCESSING
# ──────────────────────────────────────────────────────

def process_test_file(test_file: str, output_file: str, prompts_only: bool=False):
    df = pd.read_csv(test_file, sep='\t', header=None, names=["question"] )
    results = []
    for q in df['question']:
        labs = predict_labels(q)
        if prompts_only:
            p = build_prompt(q, labs)
            results.append({"question":q, "predicted_labels":"|".join(labs), "prompt":p})
        else:
            ans = answer_question(q, labs)
            results.append({"question":q, "predicted_labels":"|".join(labs), "answer":ans})
    out_df = pd.DataFrame(results)
    out_df.to_csv(output_file, index=False)
    print(f"Wrote {len(results)} rows to {output_file}")

# ──────────────────────────────────────────────────────
# 7. CLI
# ──────────────────────────────────────────────────────
if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--test-file", type=str, default="/mnt/data/Subtask3_input_test.tsv")
    parser.add_argument("--train-file", type=str, default="/mnt/data/Train_Dev.tsv")
    parser.add_argument("--output-file", type=str, default="predictions_and_answers.csv")
    parser.add_argument("--prompts-only", action="store_true", help="Only generate prompts and skip generation.")
    args = parser.parse_args()
    process_test_file(args.test_file, args.output_file, args.prompts_only)
````

````python
# -*- coding: utf-8 -*-
"""Agentic RAG pipeline for AraHealthQA Track‑1 (Sub‑task 3)
============================================================

• Predicts answer categories (information / direct_guidance / emotional_support) with a fine‑tuned classifier.
• Builds a **category‑aware prompt** that explicitly instructs Qwen‑3‑8B how to answer, according to `CATEGORY_POLICIES`.
• Adds few‑shot Q‑A pairs (label = ['3'] = emotional_support) **only** when the predicted category is *emotional_support*.
• Retrieves context from Chroma (populated with English chunks) using Qwen/Qwen3‑Embedding‑4B **filtered by similarity ≥ 0.6** after the top‑k.
• Supports two run‑modes:
    – *Full*: retrieval + generation batched on GPU.
    – *Prompts‑only*: prepare all 150 prompts for later generation.

Usage (full pipeline):
    python rag_router.py --test-file /mnt/data/Subtask3_input_test.tsv \
                         --train-file /mnt/data/Train_Dev.tsv \
                         --output-file answers.csv

Prompts‑only mode (lighter on GPU):
    python rag_router.py --prompts-only --output-file prompts.csv
"""

from __future__ import annotations

import os
import argparse
from pathlib import Path
from typing import List, Tuple, Dict, Any

import torch
import pandas as pd
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    pipeline,
)
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.docstore.document import Document

# ────────────────────────────────────────────────────────────
# 1. CONFIGURATION
# ────────────────────────────────────────────────────────────

# ‑‑ Paths & model IDs
MODEL_ID_GEN = "Qwen/Qwen3-8B"
MODEL_ID_EMB = "Qwen/Qwen3-Embedding-4B"  # retrieval embeddings
MODEL_ID_CLS = "path/to/your/classifier"  # ★ UPDATE: your fine‑tuned classifier checkpoint
CHROMA_DIR = "/content/chroma_index"      # ★ UPDATE: where Chroma persists

# ‑‑ Hardware
DEVICE = 0                 # GPU index; set -1 for CPU only

# ‑‑ Classifier settings
CLASS_LABELS = ["information", "direct_guidance", "emotional_support"]
CLASS_THRESHOLD = 0.5       # sigmoid probability threshold per label

# ‑‑ Retrieval filter
SIMILARITY_THRESHOLD = 0.6  # keep chunks with similarity ≥ 0.60 (cosine‑sim style)

# ‑‑ Category → behaviour policy
CategoryPolicy = Dict[str, Any]
CATEGORY_POLICIES: Dict[Any, CategoryPolicy] = {
    "information": {
        "style_key": "expository",
        "retrieval": "always",
        "k": 6,
    },
    "direct_guidance": {
        "style_key": "action_oriented",
        "retrieval": "hybrid",
        "k": 4,
    },
    "emotional_support": {
        "style_key": "empathetic",
        "retrieval": "avoid",
        "k": 0,
    },
    ("information", "direct_guidance"): {
        "style_key": "mix_expository_guidance",
        "retrieval": "always",
        "k": 6,
    },
    ("information", "emotional_support"): {
        "style_key": "expository_then_empathy",
        "retrieval": "always",
        "k": 6,
    },
    ("direct_guidance", "emotional_support"): {
        "style_key": "supportive_guidance",
        "retrieval": "hybrid",
        "k": 4,
    },
    ("information", "direct_guidance", "emotional_support"): {
        "style_key": "info_guidance_empathy",
        "retrieval": "always",
        "k": 6,
    },
}

# ‑‑ Style tags embedded in the prompt
STYLE_TAGS: Dict[str, str] = {
    "expository": "🔹FORMAT: factual answer · cite sources (1,2,3) · objective tone",
    "action_oriented": "🔹FORMAT: numbered steps · imperative verbs · practical guidance",
    "empathetic": "🔹FORMAT: warm, validating tone · acknowledge feelings · optimistic close",
    "mix_expository_guidance": "🔹FORMAT: facts first ↔ actionable steps second",
    "expository_then_empathy": "🔹FORMAT: factual answer then reassuring note",
    "supportive_guidance": "🔹FORMAT: gentle encouragement while giving steps",
    "info_guidance_empathy": "🔹FORMAT: facts → steps → empathy close",
}

# ‑‑ Prompt skeleton (Jinja‑like var names, rendered with `Template`)
PROMPT_TEMPLATE = r"""
{% if few_shot_block %}# Few‑Shot Emotional‑Support Examples
${few_shot_block}

{% endif %}You are **AraHealthQA Assistant**, a large‑language model helping Arabic speakers with mental‑health questions.

**User Question**
${question}

**Predicted Answer Category (multi‑label)**
${labels}

**Category‑Specific Instruction**
${category_instruction}

**Retrieval Mode**
${retrieval_instruction}

${context_block}
**Write the best possible answer, following ALL rules below**
1. Obey Arabic medical‑ethics: *no diagnoses or prescriptions* → refer to professionals.
2. Follow style guide: ${style_tag}
3. If you cite, enumerate sources as (1), (2)… matching order in CONTEXT.
4. If you lack enough info, say so briefly and suggest next steps.
5. Keep answer ≤ 200 Arabic words unless user explicitly asked for detail.
6. **Your entire answer must be written in Arabic.**
"""

# ────────────────────────────────────────────────────────────
# 2. HELPER FUNCTIONS
# ────────────────────────────────────────────────────────────

from string import Template


def needs_evidence(question: str) -> bool:
    """Heuristic: return True when guidance likely needs factual context."""
    kw = ["how", "طريقة", "خطوات", "step", "method", "best way"]
    q = question.lower()
    return any(k in q for k in kw)


def category_instruction_for(labels: List[str]) -> str:
    """Return a one‑liner instruction befitting the main category/combination."""
    single = labels[0] if len(labels) == 1 else None
    if single == "emotional_support":
        return "👉 Provide the user with empathetic emotional support."
    if single == "direct_guidance":
        return "👉 Use the retrieved context (if any) to give clear, step‑by‑step guidance."
    if single == "information":
        return "👉 Provide accurate, well‑cited information based on the context."
    # combos fall back to generic
    return "👉 Combine information, guidance, and empathy as appropriate."


def retrieval_instruction_for(mode: str) -> str:
    if mode == "always":
        return (
            "Use CONTEXT. If similarity is insufficient, say “I need more info.” Provide an informative, source‑cited answer."
        )
    if mode == "hybrid":
        return (
            "If CONTEXT (similarity ≥ 0.6) is present, ground your guidance in it; otherwise rely on your own knowledge."
        )
    # avoid
    return "Do NOT rely on external sources; answer purely from your own knowledge with emotional support."


# ────────────────────────────────────────────────────────────
# 3. MODEL INITIALISATION
# ────────────────────────────────────────────────────────────


print("[INIT] Loading retrieval embeddings & Chroma…")
emb_fn = HuggingFaceEmbeddings(model_name=MODEL_ID_EMB)
vector_db = Chroma(persist_directory=CHROMA_DIR, embedding_function=emb_fn)


class Retriever:
    """Wrapper around Chroma to apply similarity filter."""

    def __init__(self, db):
        self.db = db

    def search(self, query: str, k: int) -> List[str]:
        docs_scores = self.db.similarity_search_with_score(query, k=k)
        kept: List[str] = []
        for rank, (doc, score) in enumerate(docs_scores, 1):
            # Chroma returns "relevance" (higher is more similar)
            if score >= SIMILARITY_THRESHOLD:
                txt = doc.page_content.replace("\n", " ").strip()
                kept.append(f"({rank}) {txt}")
        return kept


retriever = Retriever(vector_db)

print("[INIT] Loading Qwen‑3‑8B for generation… (may take a while)")
bnb_cfg = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="float16",
)

if DEVICE == -1:
    raise RuntimeError("Generation on CPU is not recommended for Qwen‑3‑8B. Use --prompts-only instead.")

gen_tok = AutoTokenizer.from_pretrained(MODEL_ID_GEN, trust_remote_code=True)
gen_mod = AutoModelForCausalLM.from_pretrained(
    MODEL_ID_GEN,
    device_map="auto",
    trust_remote_code=True,
    quantization_config=bnb_cfg,
)
gen_pipe = pipeline(
    "text-generation",
    model=gen_mod,
    tokenizer=gen_tok,
    device=DEVICE,
    max_new_tokens=512,
    do_sample=True,
    temperature=0.5,
)

# ────────────────────────────────────────────────────────────
# 4. FEW‑SHOT LOADER (for emotional support)
# ────────────────────────────────────────────────────────────

FEW_SHOT_BLOCK = ""  # will be filled after CLI parse


def load_few_shot(train_file: str, n: int = 2) -> str:
    """Return n question‑answer examples with label ['3'] (emotional_support)."""
    if not os.path.isfile(train_file):
        return ""
    df = pd.read_csv(train_file, sep="\t", names=["question", "answer", "final_AS"])
    subset = df[df["final_AS"].apply(lambda x: eval(x) == ['3'])].head(n)
    pairs = subset[["question", "answer"]].to_dict("records")
    return "\n\n".join([f"Q: {p['question']}\nA: {p['answer']}" for p in pairs])


# ────────────────────────────────────────────────────────────
# 5. PIPELINE FUNCTIONS
# ────────────────────────────────────────────────────────────


def predict_labels(question: str) -> List[str]:
    tokens = cls_tok(question, return_tensors="pt", truncation=True).to(DEVICE)
    with torch.no_grad():
        logits = cls_mod(**tokens).logits.squeeze(0)
    probs = torch.sigmoid(logits).cpu().tolist()
    labels = [lab for lab, p in zip(CLASS_LABELS, probs) if p >= CLASS_THRESHOLD]
    if not labels:
        labels = [CLASS_LABELS[int(torch.argmax(logits))]]
    return labels


def build_prompt(question: str, labels: List[str]) -> str:
    key = tuple(sorted(labels))
    policy = CATEGORY_POLICIES.get(key, CATEGORY_POLICIES[labels[0]])

    # Retrieve context if policy requires
    context_lines: List[str] = []
    if policy["retrieval"] == "always" or (
        policy["retrieval"] == "hybrid" and needs_evidence(question)
    ):
        context_lines = retriever.search(question, k=policy["k"])

    context_block = (
        f"**CONTEXT (retrieved)**\n{os.linesep.join(context_lines)}" if context_lines else ""
    )

    tmpl = Template(PROMPT_TEMPLATE)
    rendered = tmpl.substitute(
        few_shot_block=FEW_SHOT_BLOCK if policy["retrieval"] == "avoid" else "",
        question=question,
        labels=", ".join(labels),
        category_instruction=category_instruction_for(labels),
        retrieval_instruction=retrieval_instruction_for(policy["retrieval"]),
        context_block=context_block,
        style_tag=STYLE_TAGS[policy["style_key"]],
    )
    return rendered


def answer_question(prompt: str) -> str:
    full = gen_pipe(prompt)[0]["generated_text"]
    return full[len(prompt) :].strip()


# ────────────────────────────────────────────────────────────
# 6. BATCH PROCESSING
# ────────────────────────────────────────────────────────────


def process_questions(test_path: str, out_path: str, prompts_only: bool = False):
    df_test = pd.read_csv(test_path, sep="\t", header=None, names=["question"])
    rows = []
    for q in df_test["question"]:
        predicted = predict_labels(q)
        prompt = build_prompt(q, predicted)
        if prompts_only:
            rows.append({
                "question": q,
                "predicted_labels": "|".join(predicted),
                "prompt": prompt,
            })
        else:
            answer = answer_question(prompt)
            rows.append({
                "question": q,
                "predicted_labels": "|".join(predicted),
                "answer": answer,
            })
    pd.DataFrame(rows).to_csv(out_path, index=False)
    print(f"[DONE] Wrote {len(rows)} rows → {out_path}")


````

````python

# ────────────────────────────────────────────────────────────
# 3. INIT RETRIEVER & GENERATOR (only if not prompts‑only)
# ────────────────────────────────────────────────────────────


def init_retriever():
    emb_fn = HuggingFaceEmbeddings(model_name=MODEL_ID_EMB)
    vectordb = Chroma(persist_directory=CHROMA_DIR, embedding_function=emb_fn)

    class _Retriever:
        def __init__(self, db):
            self.db = db

        def search(self, query: str, k: int) -> List[str]:
            docs_scores = self.db.similarity_search_with_score(query, k=k)
            kept = []
            for idx, (doc, score) in enumerate(docs_scores, 1):
                if score >= SIMILARITY_THRESHOLD:
                    text_clean = doc.page_content.replace("\n", " ").strip()
                    kept.append(f"({idx}) {text_clean}")
            return kept

    return _Retriever(vectordb)


def init_generator():
    bnb_cfg = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_use_double_quant=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype="float16",
    )
    tok = AutoTokenizer.from_pretrained(MODEL_ID_GEN, trust_remote_code=True)
    mod = AutoModelForCausalLM.from_pretrained(
        MODEL_ID_GEN,
        device_map="auto" if DEVICE >= 0 else None,
        trust_remote_code=True,
        quantization_config=bnb_cfg,
    )
    return pipeline(
        "text-generation",
        model=mod,
        tokenizer=tok,
        device=DEVICE,
        max_new_tokens=512,
        do_sample=True,
        temperature=0.5,
    )


# ────────────────────────────────────────────────────────────
# 4. FEW‑SHOT LOADER
# ────────────────────────────────────────────────────────────

FEW_SHOT_BLOCK = ""


def load_few_shot(train_file: str, n: int = 2) -> str:
    if not train_file or not os.path.isfile(train_file):
        return ""
    df = pd.read_csv(train_file, sep="\t", names=["question", "answer", "final_AS"])
    subset = df[df["final_AS"].apply(lambda x: eval(x) == ['3'])].head(n)
    return "\n\n".join([f"Q: {row.question}\nA: {row.answer}" for _, row in subset.iterrows()])


# ────────────────────────────────────────────────────────────
# 5. PROMPT BUILDER
# ────────────────────────────────────────────────────────────

retriever = None  # global retriever instance (lazy)

def build_prompt(question: str, labels: List[str]) -> str:
    global retriever
    key = tuple(sorted(labels))
    policy = CATEGORY_POLICIES.get(key, CATEGORY_POLICIES[labels[0]])

    # Retrieve context
    context_lines: List[str] = []
    if policy["retrieval"] == "always" or (
        policy["retrieval"] == "hybrid" and needs_evidence(question)
    ):
        if retriever is None:
            retriever = init_retriever()
        context_lines = retriever.search(question, k=policy["k"])

    context_block = (
        f"**CONTEXT (retrieved)**\n{os.linesep.join(context_lines)}" if context_lines else ""
    )

    tmpl = Template(PROMPT_TEMPLATE)
    return tmpl.substitute(
        few_shot_block=FEW_SHOT_BLOCK if policy["retrieval"] == "avoid" else "",
        question=question,
        labels=", ".join(labels),
        category_instruction=category_instruction_for(labels),
        retrieval_instruction=retrieval_instruction_for(policy["retrieval"]),
        context_block=context_block,
        style_tag=STYLE_TAGS[policy["style_key"]],
    )


# ────────────────────────────────────────────────────────────
# 6. MAIN PROCESS
# ────────────────────────────────────────────────────────────

generator_pipe = None  # lazy init

def process(pred_path: str, out_path: str, prompts_only: bool):
    df = pd.read_csv(pred_path, sep="\t", header=None)
    if df.shape[1] == 2:
        df.columns = ["question", "pred"]
    elif df.shape[1] == 3:
        df.columns = ["id", "question", "pred"]
    else:
        raise ValueError("Unexpected column count in predictions file")

    rows = []
    for _, row in df.iterrows():
        q = row.question
        pred_raw = str(row.pred).strip()
        # try to parse list‑like or pipe‑separated
        if pred_raw.startswith("["):
            pred_codes = [str(s).strip(" ' \" ") for s in ast.literal_eval(pred_raw)]
        else:
            pred_codes = pred_raw.split("|")
        labels = sorted({CODE2LABEL.get(code, code) for code in pred_codes})

        prompt = build_prompt(q, labels)

        if prompts_only:
            rows.append({"question": q, "predicted_labels": "|".join(labels), "prompt": prompt})
        else:
            global generator_pipe
            if generator_pipe is None:
                generator_pipe = init_generator()
            answer = generator_pipe(prompt)[0]["generated_text"][len(prompt) :].strip()
            rows.append({"question": q, "predicted_labels": "|".join(labels), "answer": answer})

    pd.DataFrame(rows).to_csv(out_path, index=False)
    print(f"[DONE] Wrote {len(rows)} rows → {out_path}")

````
